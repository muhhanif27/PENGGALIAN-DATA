```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# 1. Load data iris
iris = load_iris()
X = iris.data  # fitur (4 kolom)
y = iris.target  # label bunga
target_names = iris.target_names

# 2. Standardisasi data (biar skala fitur sama)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 3. PCA - ubah dari 4 dimensi ke 2 dimensi
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# 4. Bikin DataFrame untuk visualisasi
df_pca = pd.DataFrame(data=X_pca, columns=['PC1', 'PC2'])
df_pca['target'] = y
df_pca['species'] = df_pca['target'].apply(lambda i: target_names[i])

# 5. Visualisasi
plt.figure(figsize=(8,6))
sns.scatterplot(data=df_pca, x='PC1', y='PC2', hue='species', palette='Set1')
plt.title('PCA of Iris Dataset (2D)')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.grid(True)
plt.show()
```
