```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, GridSearchCV, cross_val_score
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Latihan_modul4.csv')
print("Dataset Info:")
print(df.info())
print("\nFirst 5 Rows:")
print(df.head())
```

```python
# Step 2: Preprocess the Data
# Check for missing values
print("\nMissing Values:")
print(df.isnull().sum())
df = df.dropna()

# Remove outliers (cap Services Revenue at 5th and 95th percentiles)
q05 = df['Services Revenue (in billion $)'].quantile(0.05)
q95 = df['Services Revenue (in billion $)'].quantile(0.95)
df = df[df['Services Revenue (in billion $)'].between(q05, q95)]
print(f"\nAfter outlier removal: {len(df)} rows remain.")
```

```python
df['Total_Sales'] = (df['iPhone Sales (in million units)'] +
                     df['iPad Sales (in million units)'] +
                     df['Mac Sales (in million units)'] +
                     df['Wearables (in million units)'])
```

```python
revenue_threshold = df['Services Revenue (in billion $)'].median()
df['High_Revenue'] = (df['Services Revenue (in billion $)'] > revenue_threshold).astype(int)
print(f"\nRevenue Threshold (median): {revenue_threshold:.2f} billion $")
print("Target variable 'High_Revenue' created.")
```
